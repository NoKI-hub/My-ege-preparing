##### № 29969 Апробация 14.05.26 (Уровень: Базовый)

Для какого наибольшего целого неотрицательного числа А выражение
- (x>A)∨(y>A)∨(x+2y<80)

истинно (т.е. принимает значение 1) при любых целых неотрицательных х и у?

```python
for a in range(1000, -1, -1):
    if all((x > a) or (y > a) or (x + 2 * y < 80) for x in range(1000) for y in range(1000)):
        print(a)
        break
```

##### № 29347 Открытый вариант 2026 (Уровень: Базовый)

На числовой прямой даны два отрезка: B = [22; 40] и C = [32; 50]. Укажите наименьшую возможную длину такого отрезка A, для которого логическое выражение

¬(x ∈ A) → ((x ∈ B) ≡ (x ∈ C))

истинно (т. е. принимает значение 1) при любом значении переменной х.

```python
def f(x):
    B = 22 <= x <= 40
    C = 32 <= x <= 50
    A = a1 <= x <= a2
    return (not A) <= (B == C)

r = []
d = [y for x in (22, 32, 40, 50) for y in (x - 0.1, x, x + 0.1)]
for a1 in d:
    for a2 in d:
        if a2 >= a1 and all(f(x) == 1 for x in d):
            r += [a2 - a1]
print(round(min(r)))
```

Ответ: 28
