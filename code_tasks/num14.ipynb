##### № 27626 Апробация 04.03.26 

![alt text](tasks/{94A2E016-B62A-4D79-B65C-2E6F1F11A043}.png)

```python
def to_any_ss(x, ss):
    r = ""
    while x > 0:
        r = str(x % ss) + r
        x = x // ss
    return r


min_z = 10000000
for x in range(2031):
    res = 6**2030 + 6**100 - x
    res6 = to_any_ss(res, 6)
    if res6.count("0") < min_z:
        min_z = res6.count("0")
print(min_z)
```

Ответ: 1930

##### № 29968 Апробация 14.05.26 (Уровень: Базовый)

![](tasks/num14_1.png)

```python
def odinn(x):
    new_x = ""
    while x > 0:
        ost = x % 11
        if ost < 10:
            new_x = str(ost) + new_x
        else:
            new_x = "A" + new_x
        x //= 11
    return new_x


a = 9 * (11**210) + 8 * (11**150)
for x in range(3000, 0, -1):
    new_a = a - x
    if odinn(new_a).count("0") == 60:
        print(x)
        break
```

Ответ: 2992

##### № 27298 (Уровень: Средний)

![](tasks/num14_2.png)

```python
from string import printable


def des(x):
    new_x = 0
    raz = len(x)
    for i in x:
        raz = raz - 1
        new_x += printable.index(i) * 40**raz
    return new_x


for x in printable[:40]:
    a1 = des("871" + x + "291")
    a2 = des("366" + x + "631")
    a3 = des("973" + x + "618")
    if (a1 + a2 + a3) % 39 == 0:
        print((a1 + a2 + a3) // 13)
        break
```

Ответ: 6461195610
