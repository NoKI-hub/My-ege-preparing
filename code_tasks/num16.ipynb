##### № 27771 Апробация 04.03.26

![alt text](tasks/{BB54596C-5A5F-47A1-8AE6-6FCF2538B658}.png)

```python
from functools import lru_cache

@lru_cache(2030)
def f(n) -> int|None:
    if n == 1: return 1
    if n > 1: return n * f(n - 1)

for i in range(1, 2030):
    f(i)
print((f(2024) - 2 * f(2023)) / f(2022))
```

Ответ: 4090506

##### № 29970 Апробация 14.05.26 (Уровень: Базовый)

![](tasks/num16_1.png)

```python
from functools import lru_cache

@lru_cache(None)
def g(n):
    if n <= 20:
        return n + 2
    else:
        return g(n - 3) + 1

def f(n):
    return 3 * g(n - 3) + 7

[g(x) for x in range(38000)]

print(f(37811))
```

Ответ: 37861

##### № 29348 Открытый вариант 2026 (Уровень: Базовый)

![](tasks/num16_2.png)

```python
from functools import lru_cache
g = 248000
@lru_cache(1000)
def f(n):
    if n < 10:
        return 1
    if n >= 10:
        return (n + 3) * f(n - 3)
    
for i in range(g):
    f(i)

print((f(247563) // 519 - 477 * f(247560)) // f(247557))
```

Ответ: 1431
