##### № 27761 Апробация 04.03.26

![alt text](tasks/image.png)

```python
from turtle import *


m = 20
tracer(0)
left(90)
down()

for _ in range(2):
    forward(1 * m)
    left(270)
    forward(16 * m)
    right(90)
up()

back(4 * m)
right(90)
fd(10 * m)
lt(90)
down()

for _ in range(2):
    fd(17 * m)
    rt(90)
    fd(7 * m)
    rt(90)
up()

for x in range(-100, 100):
    for y in range(-100, 100):
        setpos(x * m, y * m)
        dot(5, "blue")
done()
```

##### № 29959 Апробация 14.05.26 (Уровень: Базовый)

![](tasks/num6_1.png)

```python
from turtle import *

m = 10
lt(90)
speed(1000)
down()
screensize(4000, 4000)

begin_fill()
for _ in range(2):
    fd(32 * m)
    rt(90)
    fd(38 * m)
    rt(90)
end_fill()

up()
fd(25 * m)
rt(90)
fd(21 * m)
lt(90)
down()

begin_fill()
for _ in range(2):
    fd(29 * m)
    rt(90)
    bk(18 * m)
    rt(90)
end_fill()
up()
canvas = getcanvas()

cnt = 0
for x in range(-100, 100):
    for y in range(-100, 100):
        if canvas.find_overlapping(x * m, y * m, x * m, y * m) != ():
            cnt += 1
print(cnt)
done()
```

Ответ: 1705

##### № 29338 Открытый вариант 2026 (Уровень: Базовый)

![](tasks/num6_2.png)

```python
from turtle import *

m = 10
speed(1000)
down()
left(90)
right(45)

begin_fill()
for _ in range(3):
    rt(45)
    fd(10 * m)
    rt(45)
rt(315)
fd(10 * m)
rt(90)
fd(20 * m)
rt(90)
for _ in range(2):
    fd(10 * m)
    rt(90)
end_fill()
canvas = getcanvas()

up()
cnt = 0
for x in range(-50, 50):
    for y in range(-50, 50):
        x1 = x * m
        y1 = y * m
        if canvas.find_overlapping(x1, y1, x1, y1) == (5, ):
            cnt += 1

done()
print(cnt)
```
